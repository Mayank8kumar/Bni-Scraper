```python
def personal_data(url, driver):
    user_data = {}
    driver.get(url)
    time.sleep(2)
    img = driver.find_element(By.XPATH, "//div[@id='userheader']//img").get_attribute("src")
    data = driver.find_element(By.XPATH, "//div[@class='networkhometabs']")
#     user_data[l[0]] =
    
    count = 0
    for i in data.text.split("\n "):
        if count<1:
            v = i.split("\n")
            v.pop()
            
            user_data["name"] = v[-1]
            user_data["chapter"] = v[3]
            user_data["image"] = img
            count+=1
            continue

        v = i.split("\n")
        user_data[v[0]] = v[1] if len(v)>=2 else ""
#         print(v)
    return user_data
```
