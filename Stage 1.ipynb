```python
def select_region(region, driver):
    print(region)
    wait = WebDriverWait(driver, 5)
    wait.until(EC.element_to_be_clickable((By.XPATH, "//select[@id='regionId']"))).click()
    ActionChains(driver).send_keys(region).send_keys(Keys.RETURN).perform()
    
def select_chapter(chapter, driver):
    print(chapter)
    wait = WebDriverWait(driver, 5) 
    wait.until(EC.element_to_be_clickable((By.XPATH, "//select[@id='chapterName']"))).click()
#     print(chapter)
    ActionChains(driver).send_keys(chapter).send_keys(Keys.RETURN).perform()
```
